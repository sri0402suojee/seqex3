```python
import re

def load_multiple_fasta_file(fasta_file):
    """
    Reads a FASTA file and returns a dictionary with {header: sequence}.
    Cleans headers by truncating at first space or colon.
    """
    fasta_dict = {}
    with open(fasta_file, "r") as file:
        header = None
        seq = []
        for line in file:
            line = line.strip()
            if line.startswith(">"):
                if header:
                    fasta_dict[header] = ''.join(seq)
                header = line[1:]
                # Clean header (truncate at first space or colon)
                header = re.split(r'[ :]', header)[0]
                seq = []
            else:
                seq.append(line)
        if header:  # Add the last sequence
            fasta_dict[header] = ''.join(seq)
    return fasta_dict

def get_sequence_region(sequence_dict, contig_name, start_pos, end_pos):
    """
    Extracts a subsequence from a contig, adjusting positions to stay within bounds.
    Returns (subsequence, adjusted_start, adjusted_end).
    """
    full_sequence = sequence_dict[contig_name]
    seq_length = len(full_sequence)
    start = max(0, start_pos)
    end = min(seq_length, end_pos)
    return full_sequence[start:end], start, end

def reverse_complement(sequence):
    """
    Returns the reverse complement of a DNA sequence (handles lowercase).
    """
    complement = {'A':'T', 'T':'A', 'C':'G', 'G':'C', 'N':'N',
                  'a':'t', 't':'a', 'c':'g', 'g':'c', 'n':'n'}
    return ''.join(complement.get(base, 'N') for base in reversed(sequence))

def extract_and_show_sequence(fasta_path, contig_name, start_pos, end_pos, reverse_complement_flag=False):
    """
    Extracts a sequence region, shows both the original and reverse complement (if needed).
    """
    sequences = load_multiple_fasta_file(fasta_path)
    seq, start, end = get_sequence_region(sequences, contig_name, start_pos, end_pos)

    print(f"Extracted Sequence ({contig_name} {start}-{end}):")
    print(f"Original: {seq}")

    if reverse_complement_flag:
        reverse_seq = reverse_complement(seq)
        print(f"Reverse Complement: {reverse_seq}")
        return reverse_seq
    else:
        return seq

def write_fasta(results, output_path):
    """
    Writes sequences to a FASTA file with 80-character line wrapping.
    """
    with open(output_path, 'w') as file:
        for entry in results:
            file.write(entry['header'] + '\n')
            seq = entry['sequence']
            for i in range(0, len(seq), 80):
                file.write(seq[i:i+80] + '\n')

# ----------------------------
# Main Execution
# ----------------------------

# Get user inputs for FASTA file path and output file path
fasta_path = input("Enter the path to your FASTA file: ").strip()
output_filename = input("Enter the name of the output FASTA file (e.g., output.fasta): ").strip()
reverse_complement_choice = input("Reverse complement? (y/n): ").strip().lower()
reverse_complement_flag = reverse_complement_choice == 'y'

# Prepare to collect sequences
sequences = []

# Start a loop to input contig names and regions
while True:
    print("\nEnter the region as: CONTIG_NAME START END (e.g., exon-1 100 200)")
    contig_name_input = input("Enter the contig name (e.g., exon-1): ").strip()
    start_pos = int(input("Enter the start position: ").strip())
    end_pos = int(input("Enter the end position: ").strip())

    # Prepend the full contig name if necessary
    contig_name_full = f"AT1G01010.1|{contig_name_input}"

    # Extract and show sequence
    extracted_sequence = extract_and_show_sequence(fasta_path, contig_name_full, start_pos, end_pos, reverse_complement_flag)

    # Save the result
    sequences.append({
        "header": f">{contig_name_full}:{start_pos}-{end_pos}",
        "sequence": extracted_sequence
    })

    # Ask if the user wants to add more regions or finish
    more_input = input("Do you want to add another region? (y/n): ").strip().lower()
    if more_input != 'y':
        break

# Write all collected sequences to the output FASTA file
write_fasta(sequences, output_filename)
print(f"Success! {len(sequences)} sequences saved to {output_filename}.")
```
